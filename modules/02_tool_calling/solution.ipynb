The Amsterdam round trip, pointed at Istanbul. The file says: Istanbul is the only city in the world that spans two continents: Europe and Asia.


```python
messages = [
    {"role": "user", "content": "Give me a fun fact about Istanbul."}
]

response = client.chat.completions.create(
    model=model,
    messages=messages,
    tools=tools,
    max_completion_tokens=128,
    reasoning_effort="none",
)
message = response.choices[0].message
call = message.tool_calls[0]
asked_name = call.function.name

args = json.loads(call.function.arguments)
fact = get_fact(args["city"])

messages.append(message)
messages.append({"role": "tool", "tool_call_id": call.id, "content": fact})

final = client.chat.completions.create(
    model=model,
    messages=messages,
    tools=tools,
    max_completion_tokens=128,
    reasoning_effort="none",
)
final_text = final.choices[0].message.content

print(asked_name)
print(fact)
print(final_text)
```

```python
assert asked_name, "asked_name should be the tool the model requested"
assert fact, "fact should be what get_fact returned"
assert final_text, "final_text should be the last model sentence"
print("looks good")
```
