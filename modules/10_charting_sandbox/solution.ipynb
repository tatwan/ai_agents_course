Debrief only. Same kernel, after `invoices_by_year`, `run_call`, `tools`, `CHARTS`, and `client` exist.

Years: 2009–2012 have 83 invoices each, 2013 has 80.


```python
messages = [
    {
        "role": "system",
        "content": (
            "You answer with tools. Query first. Then call run_python with matplotlib code "
            "that saves the PNG. Do not invent numbers."
        ),
    },
    {
        "role": "user",
        "content": (
            "Make a bar chart of invoice count by year. "
            "Save it as invoices_by_year.png."
        ),
    },
]

for turn in range(6):
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        tools=tools,
        max_completion_tokens=500,
        reasoning_effort="none",
    )
    message = response.choices[0].message
    print("--- turn", turn + 1, "finish_reason:", response.choices[0].finish_reason, "---")
    if not message.tool_calls:
        print(message.content)
        break
    messages.append(message)
    for call in message.tool_calls:
        result = run_call(call)
        print(call.function.name, "->", str(result)[:160].replace("\n", " | "))
        messages.append({"role": "tool", "tool_call_id": call.id, "content": result})

chart_path = CHARTS / "invoices_by_year.png"
print("chart_path:", chart_path)
print("exists:", chart_path.exists())
if chart_path.exists():
    print("bytes:", chart_path.stat().st_size)
```

```python
from pathlib import Path as _Path

p = _Path(chart_path)
assert p.is_file(), "chart_path should be invoices_by_year.png under charts/"
assert p.stat().st_size > 500, "the PNG should not be an empty file"
print("looks good")
```
