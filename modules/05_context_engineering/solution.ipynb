Debrief only. Same kernel, after `FAT_SYSTEM`, `THIN_SYSTEM`, `tools`, `first_call`, and `QUESTION` exist. We only change the user text.


```python
travel_q = "Give me a fun fact about Amsterdam."

def first_tokens(system, question):
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": system},
            {"role": "user", "content": question},
        ],
        tools=tools,
        max_completion_tokens=128,
        reasoning_effort="none",
    )
    return response.usage.prompt_tokens


fat_tokens = first_tokens(FAT_SYSTEM, travel_q)
thin_tokens = first_tokens(THIN_SYSTEM, travel_q)
print("fat_tokens: ", fat_tokens)
print("thin_tokens:", thin_tokens)
print("saved:      ", fat_tokens - thin_tokens)
```

```python
assert fat_tokens > thin_tokens, "the thin first request should send fewer prompt tokens"
print("looks good")
```
