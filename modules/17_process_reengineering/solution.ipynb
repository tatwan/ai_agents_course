```python
inbox = [
    {"step": "Classify the email (billing / shipping / product).", "autonomy": "act with audit"},
    {"step": "Look up the customer in Chinook.", "autonomy": "act with audit"},
    {"step": "Issue a refund over $100.", "autonomy": "draft"},
    {"step": "Send the same 30-day reminder template.", "autonomy": "act with audit"},
    {"step": "Investigate why Q3 revenue is down.", "autonomy": "suggest"},
]
```

| Step | Autonomy | Why |
|---|---|---|
| Classify | `act with audit` | Stable labels. A classifier plus three queues is a workflow. `draft` if the labels keep moving. |
| Lookup | `act with audit` | Named tool. The facts are in Chinook. |
| Refund over $100 | `draft` | Money. A person clicks. This is module 01 row 4. |
| 30-day reminder | `act with audit` | Known steps, logged, cheap to undo. Module 01 row 1. |
| Q3 revenue | `suggest` | The next query depends on the last one. A wrong cause becomes a wrong decision. |

None of these five needed a second agent. The workshop on *their* process is next. A group that concludes it should not be automated, and can say why, is the win.

```python
EXPECTED_STEPS = [
    "Classify the email (billing / shipping / product).",
    "Look up the customer in Chinook.",
    "Issue a refund over $100.",
    "Send the same 30-day reminder template.",
    "Investigate why Q3 revenue is down.",
]
allowed = {"suggest", "draft", "act with audit", "act"}

assert isinstance(inbox, list) and len(inbox) == 5, "inbox should be the five steps"
got_steps = [row["step"] for row in inbox]
assert got_steps == EXPECTED_STEPS, "keep the step strings exactly"
assert all(row.get("autonomy") in allowed for row in inbox), "each autonomy is suggest, draft, act with audit, or act"

by_step = {row["step"]: row["autonomy"] for row in inbox}
assert by_step["Issue a refund over $100."] != "act", "money does not sit on act"
assert by_step["Send the same 30-day reminder template."] in {"act", "act with audit"}, (
    "known steps, cheap to undo -- act or act with audit, but not draft or suggest"
)
assert by_step["Investigate why Q3 revenue is down."] == "suggest", "a wrong cause becomes a wrong decision"

print(f"{'autonomy':16} step")
print("-" * 72)
for row in inbox:
    print(f"{row['autonomy']:16} {row['step']}")
```
