# 01 — Challenge solution

Debrief only. One test: who decides the next step? Rows 4 and 5 are the argument. Same six jobs as the slides. Do not treat this table as the only right sheet.

| # | Job | Shape | Who decides the next step? | How far? |
|---|-----|-------|----------------------------|----------|
| 1 | The same overdue reminder to every account past 30 days. | workflow | You. The list and the template are known. | act with audit |
| 2 | A visitor asks what jazz fusion is. | chatbot | The model talks. There is no next step. | suggest |
| 3 | A messy email with an invoice, an address change, and a question. | agent | The model. The next tool depends on what they find. | draft |
| 4 | Auto-refund up to a set limit, straight from an inbox, no one in the loop. | agent-shaped, wrong default | The model would choose. A wrong payment is expensive. | draft, or a rules workflow |
| 5 | Route each incoming request to billing, shipping, or product. | either | A three-way classifier is a workflow. An agent is for the leftovers that do not fit. | act with audit on the easy ones; draft the rest |
| 6 | Why is Q3 revenue down? | agent | The model. The next query depends on the last one. | suggest |

**1.** If you can write the steps on a slide, it is a workflow. Logged and reversible, so `act with audit`, not unbounded `act`. An agent here spends tokens to rediscover an order you already knew.

**2.** Talking is the job. No tool, no loop.

**3 and 6.** You cannot outline the path before you have seen the email or the numbers. That is the agent case. Keep a human on 6 (`suggest`) because a bad cause becomes a bad decision.

**4.** This is the one to hear out loud. Shape looks like an agent. Autonomy should not be `act`. Cost of a wrong refund is unbounded relative to the ticket. A rules workflow ("refund under $50 if the order exists") or `draft` is the grown-up answer.

**5.** Good fight. If the categories are stable, a classifier plus three queues is a workflow and you can test it. If most mail is a blob that needs two tools before you know the team, that remainder is an agent. Many rooms will split this row. That is success.
