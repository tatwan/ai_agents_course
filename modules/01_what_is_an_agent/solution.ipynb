# 01 — Challenge solution

Debrief only. One test: who decides the next step? Rows 7 and 8 are the argument. Do not treat this table as the only right sheet.

| # | Job | Shape | Who decides the next step? | How far? |
|---|-----|-------|----------------------------|----------|
| 1 | Same overdue reminder to every 30-day account. | workflow | You. The list and the template are known. | act |
| 2 | Visitor asks what jazz fusion is. | chatbot | The model talks. There is no next step. | suggest |
| 3 | Messy email: invoice, address, then a recommendation. | agent | The model. The next tool depends on what they find. | draft |
| 4 | Helena: count, then rep, always that order. | workflow | You. That is the cell we already ran. | act |
| 5 | Nightly invoice CSV to a file share. | workflow | You. Same steps every night. | act |
| 6 | Why is Q3 revenue down? | agent | The model. The next query depends on the last one. | suggest |
| 7 | Auto-refund up to $5,000 from email, no one in the loop. | agent-shaped, wrong default | The model would choose. A wrong payment is expensive. | draft, or a rules workflow |
| 8 | Route email to billing, shipping, or product. | either | A three-way classifier is a workflow. An agent is for the leftovers that do not fit. | act on the easy ones; draft the rest |

**1, 4, 5.** If you can write the steps on a slide, it is a workflow. An agent here spends tokens to rediscover an order you already knew.

**2.** Talking is the job. No tool, no loop.

**3 and 6.** You cannot outline the path before you have seen the email or the numbers. That is the agent case. Keep a human on 6 (`suggest`) because a bad cause becomes a bad decision.

**7.** This is the one to hear out loud. Shape looks like an agent. Autonomy should not be `act`. Cost of a wrong refund is unbounded relative to the ticket. A rules workflow ("refund under $50 if the order exists") or `draft` is the grown-up answer.

**8.** Good fight. If the categories are stable, a classifier plus three queues is a workflow and you can test it. If most mail is a blob that needs two tools before you know the team, that remainder is an agent. Many rooms will split this row. That is success.
