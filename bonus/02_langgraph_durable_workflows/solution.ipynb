```python
challenge_config = {'configurable': {'thread_id': 'bonus-case-201'}}
challenge_case = new_case('case-201', amount_usd=800, days=45, opened=True)

challenge_paused = refund_graph.invoke(challenge_case, challenge_config)
challenge_payload = challenge_paused['__interrupt__'][0].value
challenge_snapshot = refund_graph.get_state(challenge_config)
challenge_result = refund_graph.invoke(
    Command(resume='reject'),
    challenge_config,
)

print('payload:', challenge_payload)
print('paused next:', challenge_snapshot.next)
print('decision:', challenge_result['decision'])
print('audit:', challenge_result['audit_log'])
```
